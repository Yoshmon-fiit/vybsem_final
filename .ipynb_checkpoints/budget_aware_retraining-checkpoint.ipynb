# Does pacing a retraining budget improve stream classification?

**Question.** When a deployed classifier may be replaced at most *B* times, should the evidence required for a
replacement depend on how much budget is left? Each *pacing* rule is compared with the identical rule whose
pacing multiplier is fixed at one. Every comparison is paired within the same stream. The result is negative.

**Setup.** One Hoeffding tree (River 0.23), frozen between replacements. ADWIN watches the prediction errors and
a decision is taken every 200 samples. After an accepted alarm the candidate is fitted on 400 labels that arrive
*after* the alarm and validated on the next 200. The budget counts deployments. Streams are SEA with repeated
*minor* (threshold 8↔9) and *major* (label reversal) changes, plus the Elec2 electricity series.

Run the cells top to bottom; about ten minutes on one CPU (raise `JOBS` for more). Elec2 is downloaded from a
public GitHub copy with six features and no date/day columns.

```python
import itertools, os, urllib.request
from itertools import product
import numpy as np
import pandas as pd
from scipy import stats
from scipy.spatial.distance import jensenshannon
from joblib import Parallel, delayed
from river import datasets, drift, tree

STEP = VALID = 200            # decision grid and validation block (samples)
W = 400                       # fitting window
HISTORY, LENGTH = 1200, 24000
BUDGETS = [2, 5, 6, 8, 11, 12]
EVAL_SEEDS, TUNE_SEEDS = list(range(31, 61)), [11, 12, 13]
JOBS = 1                      # raise to use several CPU cores
LOW, HIGH = 0.5, 2.5          # clip range of the pacing multiplier
METHODS = ["Pacing", "Fixed", "Gain pacing", "Gain fixed", "Queue", "Severity",
           "Label sufficiency", "Greedy", "Frozen", "Incremental", "Offline best"]
GRID = {"Pacing": [.5, 1, 2, 4], "Gain pacing": [.5, 1, 2, 4], "Queue": [4, 8, 16, 24],
        "Severity": [.05, .1, .2, .3], "Label sufficiency": [2, 4, 8, 16]}
CASES = [(o, k, 24, LENGTH) for o in ("Minor first", "Major first", "Interleaved") for k in ("Abrupt", "Gradual")]
CASES += [("Interleaved", k, 24, 2 * LENGTH) for k in ("Abrupt", "Gradual")]      # slower streams
```

## Data
Synthetic streams have 24 evenly spaced changes. *Minor first*, *Major first* and *Interleaved* set the order of
minor and major changes; *Gradual* changes take 400 samples; the two slow conditions double the length to 48,000
samples, so changes are about 1,920 rather than 960 samples apart. Elec2 keeps its chronological order: the first
20% selects settings, the remaining 80% is scored once.

```python
def sea(seed, order, kind, count, length):
    """SEA stream with `count` evenly spaced changes. A minor change toggles the threshold 8<->9,
    a major change reverses the labels. Returns X, y, the change flags and each sample's phase."""
    centers = np.rint(np.arange(1, count + 1) / (count + 1) * length).astype(int)
    major = np.arange(count) % 2 == 1 if order == "Interleaved" else (np.arange(count) >= count // 2) ^ (order == "Major first")
    data = list(itertools.islice(datasets.synth.SEA(variant=0, seed=seed), HISTORY + length))
    X = np.array([list(x.values()) for x, _ in data])
    y = np.array([lab for _, lab in data], dtype=bool)
    u = np.random.default_rng(seed + 1000).random(len(X))
    t = np.arange(len(X)) - HISTORY + 1
    phase = np.zeros(len(X), int)
    for c in centers:                                   # gradual: a 400-sample probabilistic transition
        phase += (u < np.clip(.5 + (t - c) / 400, 0, 1)) if kind == "Gradual" else (t >= c)
    threshold, reverse = 8, False
    for j, strong in enumerate(major):
        reverse ^= bool(strong)
        threshold = threshold if strong else 17 - threshold
        m = phase == j + 1
        y[m] = (X[m, 0] + X[m, 1] > threshold) ^ reverse
    return dict(X=X, y=y, major=major, phase=phase)


ELEC_URL = "https://raw.githubusercontent.com/scikit-multiflow/streaming-datasets/master/elec.csv"

def electricity():
    """Elec2 in chronological order: first 20% for tuning, the rest scored once."""
    if not os.path.exists("elec.csv"):
        urllib.request.urlretrieve(ELEC_URL, "elec.csv")
    d = pd.read_csv("elec.csv")
    X, y = d.drop(columns="class").to_numpy(float), d["class"].to_numpy(bool)
    k = len(y) // 5
    cut = lambda a, b: dict(X=X[a:b], y=y[a:b], major=None, phase=None)
    return cut(0, k), cut(k - HISTORY, len(y))
```

## Learner and candidate bank
The candidate deployed at block *b* is fitted on 400 labels, validated on the next 200 and then scored. Because
the tree is frozen, its predictions can be computed once for the whole stream and reused by every policy, which
also makes the hindsight optimum exact on this grid.

```python
def new_tree():
    return tree.HoeffdingTreeClassifier(grace_period=30, delta=0.01, tau=0.10, leaf_prediction="mc")

def batch_prob(model, X):
    """P(label=True) of a frozen numeric tree for all rows (same leaves as predict_one in River 0.23)."""
    p = np.empty(len(X))
    def visit(node, idx):
        if len(idx):
            if hasattr(node, "children"):
                left = X[idx, node.feature] <= node.threshold
                visit(node.children[0], idx[left]); visit(node.children[1], idx[~left])
            else:
                p[idx] = node.prediction({}, tree=model).get(True, 0)
    visit(model._root, np.arange(len(X)))
    return p


class Stream:
    """One stream plus a cache of the candidate model that would be deployed at each 200-sample block."""
    def __init__(self, s, fresh=True):
        self.X, self.y, self.major, self.phase, self.fresh = s["X"], s["y"], s["major"], s["phase"], fresh
        self.n = len(self.y) - HISTORY
        self.ends = np.append(np.arange(0, self.n, STEP), self.n)
        self.nb = len(self.ends) - 1
        self.maps = [dict(enumerate(r)) for r in self.X]
        self.bank, self.alarm_cache, self.suff = {}, {}, {}

    def cand(self, b):
        """Model deployed at block b: fitted on W labels, validated on the next VALID labels, then scored."""
        if b not in self.bank:
            fe = HISTORY + self.ends[b] - VALID
            m = new_tree()
            for i in range(fe - W, fe):
                m.learn_one(self.maps[i], bool(self.y[i]))
            p = batch_prob(m, self.X[fe:])
            truth = self.y[fe:]
            err, loss = (p > .5) != truth, np.where(truth, 1 - p, p)
            cost = np.r_[0, np.cumsum(err[VALID:], dtype=int)]
            self.bank[b] = (err[:VALID], err[VALID:], loss[:VALID], loss[VALID:], cost)
        return self.bank[b]

    def alarms(self, b):
        """ADWIN alarms (block, sample) of the model deployed at block b, computed lazily."""
        if b not in self.alarm_cache:
            def scan():
                det = drift.ADWIN(delta=0.01)
                for k, e in enumerate(self.cand(b)[1], 1):
                    det.update(int(e))
                    if det.drift_detected:
                        yield (self.ends[b] + k + STEP - 1) // STEP, self.ends[b] + k
            self.alarm_cache[b] = ([], scan())
        seen, gen = self.alarm_cache[b]
        i = 0
        while True:
            if i == len(seen):
                nxt = next(gen, None)
                if nxt is None:
                    return
                seen.append(nxt)
            yield seen[i]
            i += 1

    def errors(self, events):
        """Prediction errors of the schedule that deploys candidates at `events`."""
        parts, a = [], 0
        for z in list(events) + [self.nb]:
            parts.append(self.cand(a)[1][:self.ends[z] - self.ends[a]])
            a = z
        return np.concatenate(parts)
```

## Policies
* **Pacing / Fixed** (degradation gate). Accept an alarm when the error on the last 200 samples exceeds the error
  on the current model's validation block by more than *z·m·SE*. Pacing sets *m = clip(1/ρ, 0.5, 2.5)*, where ρ is
  the remaining budget fraction divided by the remaining stream fraction; Fixed sets *m = 1*.
* **Gain pacing / Gain fixed**. After waiting for fresh labels, accept the candidate when its mean error reduction
  over the incumbent, measured on the same 200 labels, exceeds *z·m·SE*. A rejected fit costs compute, not budget.
* **Queue, Severity, Label sufficiency**. Simplified analogues, not reproductions, of RCCDA (a spending-debt
  virtual queue), severity-aware adaptation (shift of the loss distribution) and CALIPER (a kernel-regression
  sufficiency test; the original is label-free, this version predicts labels).
* **Greedy** accepts every alarm until the cap, **Frozen** never retrains, **Incremental** learns from every label
  with no cap, and **Offline best** is the hindsight-optimal schedule of at most *B* deployments on the same grid.

```python
def rho(S, budget, used, block):
    """Pacing ratio: remaining budget fraction / remaining stream fraction."""
    return ((budget - used) / budget) / ((S.n - S.ends[block]) / S.n)

def accepts(S, method, z, start, block, used, budget, debt):
    """Decision at an alarm for the degradation, queue and severity rules (others accept every alarm)."""
    ref, errs, base_loss, loss, _ = S.cand(start)
    end = S.ends[block] - S.ends[start]
    recent = errs[end - STEP:end]
    if method in ("Pacing", "Fixed"):
        a, b = (recent.sum() + 1) / (len(recent) + 2), (ref.sum() + 1) / (len(ref) + 2)
        se = np.sqrt(a * (1 - a) / len(recent) + b * (1 - b) / len(ref))
        m = np.clip(1 / rho(S, budget, used, block), LOW, HIGH) if method == "Pacing" else 1
        return a - b > z * m * se
    if method == "Queue":                         # error growth vs. accumulated spending debt
        best = min(ref.mean(), errs[:end].reshape(-1, STEP).mean(axis=1).min())
        return z * (recent.mean() - best) >= max(0, debt - (end - 1) * budget / S.n) + .5 - budget / S.n
    if method == "Severity":                      # shift of the loss distribution
        new = loss[end - STEP:end]
        if new.mean() <= base_loss.mean():
            return False
        h = [np.histogram(v, np.linspace(0, 1, 9))[0] for v in (base_loss, new)]
        score = (.4 * stats.ks_2samp(base_loss, new, method="asymp").statistic
                 + .3 * stats.wasserstein_distance(base_loss, new) + .3 * jensenshannon(*h, base=2) ** 2)
        return score >= z
    return True

def sufficient(S, first, now, ess_min):
    """Label-sufficiency proxy: kernel predictions of the last 32 labels must improve with locality."""
    key = (first, now, ess_min)
    if key not in S.suff:
        end = HISTORY + S.ends[now]
        begin = max(HISTORY + S.ends[first], end - W - VALID)
        X, t = S.X[begin:end], S.y[begin:end, None].astype(float)
        train, query = X[:-32], X[-32:]
        d = np.linalg.norm((query[:, None] - train) / np.maximum(train.std(axis=0), 1e-7), axis=2)
        d /= np.maximum(d.mean(axis=1, keepdims=True), 1e-8)
        losses, ok = [], True
        for loc in (0, 1, 2):
            w = np.exp(-loc * d)
            if (w.sum(1) ** 2 / (w * w).sum(1)).min() < ess_min:
                ok = False
                break
            losses.append(np.mean(((w @ t[:-32]) / w.sum(1, keepdims=True) - t[-32:]) ** 2))
        S.suff[key] = ok and losses[0] >= losses[1] >= losses[2]
    return S.suff[key]

def run_policy(S, method, budget, z=None, cap=None):
    """Simulate one policy; returns the deployment blocks and the outcomes of its gate checks."""
    events, gates, start, debt, busy = [], [], 0, 0., 0
    while len(events) < (budget if cap is None else cap):
        nxt = None
        for block, _ in S.alarms(start):
            if block >= S.nb:
                break
            if S.ends[block] <= busy:
                continue
            ok = accepts(S, method, z, start, block, len(events), budget, debt)
            if method in ("Pacing", "Fixed"):
                gates.append(ok)
            if not ok:
                continue
            ready = block + ((W + VALID + STEP - 1) // STEP if S.fresh else 0)   # wait for post-alarm labels
            if method == "Label sufficiency":
                ready = max(ready, block + 1)
                while ready < S.nb and not sufficient(S, block, ready, z):
                    ready += 1
            if method in ("Gain pacing", "Gain fixed") and ready < S.nb:
                end = S.ends[ready] - S.ends[start]
                gain = S.cand(start)[1][end - VALID:end].astype(float) - S.cand(ready)[0]
                se = max(gain.std(ddof=1) / np.sqrt(VALID), 1 / VALID)
                m = np.clip(1 / rho(S, budget, len(events), ready), LOW, HIGH) if method == "Gain pacing" else 1
                gates.append(gain.mean() > z * m * se)
                if not gates[-1]:
                    busy = S.ends[ready]
                    continue
            nxt = ready if ready < S.nb else None
            break
        if nxt is None:
            break
        debt = max(0, max(0, debt - (S.ends[nxt] - S.ends[start] - 1) * budget / S.n) + 1 - budget / S.n)
        events.append(nxt); start = nxt
    return events, gates

def offline_best(S, budgets):
    """Hindsight-optimal schedules with at most B deployments on the same grid (dynamic programming)."""
    n = S.nb
    edge = np.full((n + 1, n + 1), np.inf)
    for s in range(n):
        edge[s, s + 1:] = S.cand(s)[4][S.ends[s + 1:] - S.ends[s]]
    cost, paths, out = edge[0].copy(), [[] for _ in range(n + 1)], {}
    for b in range(1, max(budgets) + 1):
        prop = cost[:, None] + edge
        src, prev = prop.argmin(axis=0), list(paths)
        for j in range(1, n + 1):
            if prop[src[j], j] < cost[j]:
                cost[j], paths[j] = prop[src[j], j], prev[src[j]] + [int(src[j])]
        if b in budgets:
            out[b] = list(paths[n])
    return out

def incremental(S):
    m = new_tree()
    for x, y in zip(S.maps[:HISTORY], S.y[:HISTORY]):
        m.learn_one(x, bool(y))
    err = []
    for x, y in zip(S.maps[HISTORY:], S.y[HISTORY:]):
        err.append(m.predict_one(x) != y)
        m.learn_one(x, bool(y))
    return np.array(err)
```

## Experiment

```python
def run_stream(stream, info, params, budgets=BUDGETS, methods=METHODS, fresh=True, keep_err=False):
    """Run every policy at every budget on one realized stream and return one row per combination."""
    S = Stream(stream, fresh)
    best = offline_best(S, budgets) if "Offline best" in methods else {}
    inc = incremental(S) if "Incremental" in methods else None
    rows = []
    for b, m in product(budgets, methods):
        gates, z = [], params.get({"Fixed": "Pacing", "Gain fixed": "Gain pacing"}.get(m, m))
        if m == "Offline best": ev = best[b]
        elif m in ("Frozen", "Incremental"): ev = []
        else: ev, gates = run_policy(S, m, b, z)
        err = inc if m == "Incremental" else S.errors(ev)
        row = dict(**info, method=m, budget=b, acc=100 * (1 - err.mean()), used=len(ev),
                   events=tuple(int(S.ends[e]) for e in ev), tested=len(gates), accepted=sum(gates))
        if m == "Greedy":                                  # deployments an uncapped greedy rule would make
            row["demand"] = len(run_policy(S, m, b, cap=S.nb)[0])
        if S.phase is not None and ev:                     # share of fit labels from the regime deployed into
            row["purity"] = np.mean([np.mean(S.phase[HISTORY + S.ends[e] - VALID - W:HISTORY + S.ends[e] - VALID]
                                             == S.phase[HISTORY + S.ends[e]]) for e in ev])
        if info.get("order") == "Major first" and info["kind"] == "Abrupt" and ev:
            t = S.ends[ev[-1]]                             # label polarity after the last replacement
            pol = np.r_[False, np.cumsum(S.major) % 2 == 1][S.phase]
            fit_pol = pol[HISTORY + t - VALID - W:HISTORY + t - VALID].mean() > .5
            row.update(opposite=np.mean(pol[HISTORY + t:] != fit_pol), tail_acc=100 * (1 - err[t:].mean()))
        if keep_err:
            row["err"] = err
        rows.append(row)
    return rows

def tune(streams, budgets=(2, 6, 12)):
    """Pick each rule's setting by mean accuracy on separate tuning streams."""
    recs = []
    for s in streams:
        S = Stream(s)
        for m, grid in GRID.items():
            for z, b in product(grid, budgets):
                recs.append((m, z, 100 * (1 - S.errors(run_policy(S, m, b, z)[0]).mean())))
    t = pd.DataFrame(recs, columns=["method", "z", "acc"]).groupby(["method", "z"]).acc.mean()
    return {m: float(t[m].idxmax()) for m in GRID}, t
```

## Sanity check
The batch predictions of the frozen tree must equal River's `predict_one`.

```python
s = sea(31, "Interleaved", "Abrupt", 24, LENGTH)
m = new_tree()
for i in range(600, 1000):
    m.learn_one(dict(enumerate(s["X"][i])), bool(s["y"][i]))
sample = s["X"][1000:3000]
assert np.array_equal(batch_prob(m, sample) > .5,
                      np.array([bool(m.predict_one(dict(enumerate(x)))) for x in sample]))
print("batch predictions match River")
```

## Tuning and runs
Each rule gets four candidate settings, chosen by mean accuracy on three separate tuning seeds at budgets 2, 6 and
12. The evaluation seeds are disjoint from them. A setting at the edge of its grid is worth flagging, because the
search may have been too narrow.

```python
train = [sea(s, o, k, 24, LENGTH) for s in TUNE_SEEDS
         for o, k in [("Minor first", "Abrupt"), ("Major first", "Gradual"), ("Interleaved", "Abrupt")]]
params, tuning = tune(train)
prefix, suffix = electricity()
real_params, real_tuning = tune([prefix])
print("synthetic settings:", params, "edge picks:", [m for m, z in params.items() if z in (min(GRID[m]), max(GRID[m]))])
print("Elec2 settings:    ", real_params, "edge picks:", [m for m, z in real_params.items() if z in (min(GRID[m]), max(GRID[m]))])
```

```python
def one_seed(seed):
    rows = []
    for o, k, c, L in CASES:
        rows += run_stream(sea(seed, o, k, c, L), dict(order=o, kind=k, count=c, length=L, seed=seed), params)
    for k in ("Abrupt", "Gradual"):      # ablation: fit at once instead of waiting for post-alarm labels
        rows += run_stream(sea(seed, "Interleaved", k, 24, LENGTH),
                           dict(order="Interleaved", kind=k, count=24, length=LENGTH, seed=seed, variant="Immediate"),
                           params, [6], ["Fixed", "Greedy"], fresh=False)
    return rows

res = pd.DataFrame([r for part in Parallel(n_jobs=JOBS)(delayed(one_seed)(s) for s in EVAL_SEEDS) for r in part])
imm, syn = res[res.variant.eq("Immediate")], res[res.variant.ne("Immediate")].drop(columns="variant")
real = pd.DataFrame(run_stream(suffix, dict(order="Real", kind="Electricity", count=0,
                                            length=len(suffix["y"]) - HISTORY), real_params, keep_err=True))
print(f"{len(CASES)} conditions x {res.seed.nunique()} seeds; {len(suffix['y']) - HISTORY:,} Elec2 predictions")
```

## Results

```python
KEY = ["order", "kind", "count", "length", "budget"]

def boot_ci(a, reps=1500):
    """Percentile bootstrap interval for a mean, resampling whole seeds."""
    a = np.asarray(a, float)
    idx = np.random.default_rng(0).integers(0, len(a), (reps, len(a)))
    return np.quantile(a[idx].mean(axis=1), [.025, .975])

def series_ci(err, block=800, reps=1000):
    """Moving-block bootstrap interval for one chronological error sequence."""
    rng, n = np.random.default_rng(1), len(err)
    pre = np.r_[0, np.cumsum(err)]
    starts = rng.integers(0, n - block + 1, (reps, n // block))
    miss = (pre[starts + block] - pre[starts]).sum(axis=1)
    return np.quantile(100 * (1 - miss / (block * (n // block))), [.025, .975])

acc = syn.pivot_table(index=KEY + ["seed"], columns="method", values="acc")
assert (acc["Offline best"] >= acc.drop(columns=["Incremental", "Offline best"]).max(axis=1) - 1e-9).all()
assert (syn.used <= syn.budget).all()
print(f"{len(syn):,} policy-budget rows; hindsight bound and budget caps hold")
```

### Table 1. Does fresh training data matter?
Budget 6 on 24 interleaved changes. *Fresh* waits for 400 post-alarm labels before fitting; *Immediate* fits on the
400 labels that precede the alarm boundary. Purity is the share of fitting labels drawn from the regime the model
is deployed into. Waiting also changes the later alarms, so the number of replacements is not held fixed and is
reported alongside.

```python
fresh = syn[(syn.order == "Interleaved") & (syn["count"] == 24) & (syn.length == LENGTH)
            & (syn.budget == 6) & syn.method.isin(["Fixed", "Greedy"])].assign(variant="Fresh")
both = pd.concat([fresh, imm])
g = both.groupby(["kind", "variant", "method"])[["acc", "used", "purity"]].mean()
table1 = pd.DataFrame({"Fixed acc": g.xs("Fixed", level=2).acc, "Greedy acc": g.xs("Greedy", level=2).acc,
                       "Replacements": g.xs("Fixed", level=2).used,
                       "Purity": g.xs("Fixed", level=2).purity}).round(3)
display(table1)
d = both[both.method == "Fixed"].pivot_table(index=["kind", "seed"], columns="variant", values="acc")
for k in ("Abrupt", "Gradual"):
    x = (d.Fresh - d.Immediate)[k]
    print(f"{k}: fresh minus immediate = {x.mean():+.2f} pp, 95% CI {boot_ci(x).round(2)}")
```

### Table 2. Does the cap bind, and does pacing change the decision?
24 abrupt changes: three severity orders x all seeds per row. *Demand* is the share of streams on which an uncapped
greedy rule would use more than *B* deployments, so a high value means the cap really binds. *Changed* counts the
streams on which pacing deploys at different times from its fixed control. Gaps are mean paired accuracy
differences in percentage points, with bootstrap intervals over seeds and win/tie/loss counts.

```python
core = syn[(syn.kind == "Abrupt") & (syn.length == LENGTH)].assign(ev=lambda d: d.events.astype(str))
rows = []
for b, g in core.groupby("budget"):
    a = g.pivot_table(index=["order", "seed"], columns="method", values="acc")
    ev = g.pivot_table(index=["order", "seed"], columns="method", values="ev", aggfunc="first")
    out = {"B": b, "Demand > B": f"{100 * (g[g.method == 'Greedy'].demand > b).mean():.0f}%",
           "Changed": f"{(ev.Pacing != ev.Fixed).sum()}/{len(ev)}"}
    for m, c in (("Pacing", "Fixed"), ("Gain pacing", "Gain fixed")):
        diff = (a[m] - a[c]).round(8)
        lo, hi = boot_ci(diff.groupby("seed").mean())
        out[f"{m} gap"] = round(diff.mean(), 3)
        out[f"{m} CI"] = f"[{lo:+.3f}, {hi:+.3f}]"
        out[f"{m} W/T/L"] = f"{(diff > 0).sum()}/{(diff == 0).sum()}/{(diff < 0).sum()}"
    rows.append(out)
table2 = pd.DataFrame(rows)
display(table2)
for m, c in (("Pacing", "Fixed"), ("Gain pacing", "Gain fixed")):
    d = (acc[m] - acc[c]).groupby("seed").mean()
    s = syn[syn.method == m]
    print(f"{m} minus {c}: {d.mean():+.3f} pp, 95% CI {boot_ci(d).round(3)} over "
          f"{acc.groupby(KEY).ngroups} condition-budget combinations; "
          f"the gate accepted {s.accepted.sum():,} of {s.tested.sum():,} checks "
          f"({100 * s.accepted.sum() / s.tested.sum():.1f}%)")
```

### Table 3. Whole-stream accuracy (%)
Each synthetic column is one budget, averaged over the eight conditions and all seeds with equal weight. The last
column is the single Elec2 test series at budget 6. *Persistence* predicts the previous label and *Majority* the
more frequent class of the tuning prefix; both are reference forecasts, and Offline best is a hindsight bound
rather than a policy.

```python
order = ["Pacing", "Fixed", "Gain pacing", "Gain fixed", "Queue", "Severity", "Label sufficiency",
         "Greedy", "Frozen", "Incremental", "Offline best"]
table3 = syn.pivot_table(index="method", columns="budget", values="acc").reindex(order)
table3["Elec2 (B=6)"] = real[real.budget == 6].set_index("method").acc
y_test = suffix["y"][HISTORY:]
table3.loc["Persistence", "Elec2 (B=6)"] = 100 * np.mean(y_test == suffix["y"][HISTORY - 1:-1])
table3.loc["Majority", "Elec2 (B=6)"] = 100 * np.mean(y_test == (prefix["y"].mean() > .5))
display(table3.round(2).fillna("—"))
e = real[real.budget == 6].set_index("method")
ci = e.err.map(series_ci)
display(pd.DataFrame({"Elec2 acc": e.acc, "low": ci.str[0], "high": ci.str[1]}).loc[order[:-1]].round(2))
```

### Slow versus fast changes
Budget 6, 24 interleaved changes, 24,000 samples (changes ≈960 apart) against 48,000 (≈1,920 apart). Mean paired
differences in percentage points: retraining pays off more when changes are further apart, but the pacing
multiplier still makes almost no difference.

```python
il = acc.xs(6, level="budget").reset_index().query("order == 'Interleaved'")
il["Pacing − Frozen"] = il.Pacing - il.Frozen
il["Pacing − Fixed"] = il.Pacing - il.Fixed
il["Gain pacing − Gain fixed"] = il["Gain pacing"] - il["Gain fixed"]
display(il.groupby(["kind", "length"])[["Pacing − Frozen", "Pacing − Fixed",
                                        "Gain pacing − Gain fixed"]].mean().round(3))
```

### What happens after the last replacement? (major-first abrupt streams)
*Opposite* is the share of the remaining stream whose label polarity differs from that of the final model's
fitting window, and *tail_acc* is accuracy after the last replacement. This diagnoses the budgets at which overall
accuracy collapses.

```python
mf = syn[(syn.order == "Major first") & (syn.kind == "Abrupt") & syn.method.isin(["Fixed", "Queue"])]
display(mf.groupby(["method", "budget"])[["acc", "opposite", "tail_acc"]].mean().round(3).unstack(0))
```

## What the experiment supports
* The pacing multiplier almost never changed a deployment decision. The gates accepted nearly every alarm that
  reached them (Table 2), so pacing and its fixed control behave alike, and the accuracy gap is indistinguishable
  from zero.
* The age of the fitting data matters more than the acceptance threshold (Table 1).
* Spending the budget early can leave a model in the wrong state for the rest of a stream (last table), and
  hindsight scheduling leaves several points of accuracy unused (Table 3).

**Limits.** One learner, frozen between replacements; prompt labels; a known horizon; cost counted as deployments,
although rejected gain fits also consume compute; simplified analogues of published rules; settings chosen from
four values, sometimes at the edge of the grid; and a single real series on which predicting the previous label
beats every policy. The synthetic label reversals extend standard SEA and create parity effects at some budgets.
All results describe the tested streams.
